Comments in Python
Comments are notes written inside code for humans.

Python ignores ordinary comments during execution.

The comment explains what the following code is doing.
Important rules
- A comment starts with #.
- Comments are not executed as Python statements.
- Good comments explain intention, not obvious syntax.
- Avoid unnecessary comments on every line.
